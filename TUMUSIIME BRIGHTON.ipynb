## q2: 4 marks

```python
# Your answer here
def percent_error(measured, expected):
    return 100.0 * (measured - expected) / expected
```

## q3: 3 marks
Write slab_mass(area_m2, thickness_m) returning the mass of the concrete slab in kg, taking concrete as 2400 kg/m3.

```python
# Your answer here
def slab_mass(area_m2, thickness_m):
    density_concrete = 2400.0  # kg/m^3
    volume = area_m2 * thickness_m
    return volume * density_concrete
```

## q4: 5 marks
Write efficiency(out_w, in_w) returning the percentage, and returning None when in_w is zero or negative.

```python
# Your answer here
def efficiency(out_w, in_w):
    if in_w <= 0:
        return None
    return (out_w / in_w) * 100.0
```

## q5: 5 marks
READINGS is given in the notebook. Using your own percent_error against a nominal of 100, set worst_reading to the reading with the largest error by size, and worst_error to that reading's signed percent error.

You may compare these six readings directly; a loop is not required this week.

```python
READINGS = [98.5, 103.2, 99.0, 111.4, 94.8, 100.6]

# Compute percent error for each reading against a nominal value of 100
errors = [percent_error(r, 100.0) for r in READINGS]

# Find the reading with the largest absolute error
abs_errors = [abs(e) for e in errors]
max_err_idx = abs_errors.index(max(abs_errors))

worst_reading = READINGS[max_err_idx]
worst_error = errors[max_err_idx]
```
